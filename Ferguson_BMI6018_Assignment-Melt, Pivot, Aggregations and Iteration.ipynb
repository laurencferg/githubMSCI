```python
#%% Libraries
import pandas as pd
import numpy as np

#%% Importing Data
heart_data = pd.read_csv("arrhythmia.csv", na_values="?")
heart_data.head(5)
```

```python
# Melt
intervals = ["pr_interval", "qt_interval","t_interval","p_interval"] #isolate the columns I want to see in the melt
df_melt = pd.melt(heart_data, id_vars=['weight_kg'], value_vars = intervals, var_name="interval",value_name="msec") #set weight
#as the primary variable, then let the intervals and thier values stack up in the next columns, title those coulmns
df_melt.head(10) #show a sample
```

```python
# Pivot
df_pivot = heart_data.pivot(values="qt_interval",columns="sex") #pivot so each sex gets a column with
#qt intervals listed, the table index stays the index
df_pivot.head(5) #display a sample
```

```python
# Aggregation
heart_aggregate = heart_data['qrs_duration'].agg(['mean', 'min', 'max']) #run the mean, minimum, and maximum aggregate statistics
#for the qrs_duration column
print(heart_aggregate)
```

```python
# Iteration
for row in heart_data.itertuples(): #iterate through each row in the dataframe
    if row.qt_interval > 440 and row.weight_kg > 70: #create a logic condition for the following code
        print(f"Patient {row.Index}: age {row.age}, QT {row.qt_interval}") #print the desired information if the above is satisfied
```

```python
# Groupby
heart_groupby = heart_data.groupby(["sex"]).mean() #run the mean method for each column and differentiate by the group "sex"
print(heart_groupby)
```

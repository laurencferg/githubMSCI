```python
# -*- coding: utf-8 -*-

#%% Libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

#%% Importing Data
flights_data = pd.read_csv('flights.csv')
flights_data.head(10)
weather_data_pd = pd.read_csv('weather.csv')
weather_data_np = weather_data_pd.to_numpy()
```

```python
#%% Pandas Data Filtering/Sorting Question Answering
#use flights_data
#Question 1 How many flights were there from JFK to SLC? Int
q_1 = ((flights_data['origin'] == 'JFK') & (flights_data['dest'] == 'SLC')).sum() #sum the instances where both are true, origin of JFK and destination of SLC
print(q_1)
```

```python
#Question 2 How many airlines fly to SLC? Should be int
q_2 = flights_data.loc[flights_data['dest'] == 'SLC', 'carrier'].nunique() #select rows where the dest is SLC and find unique occurances in the carrier column of that row
print(q_2)
```

```python
#Question 3 What is the average arrival delay for flights to RDU? float
q_3 = float(flights_data.loc[flights_data['dest'] == 'RDU', 'arr_delay'].mean()) #select the rows where destination is RDU and for the value in the arrival delay column, average them
print(q_3)
```

```python
#Question 4 What proportion of flights to SEA come from the two NYC airports (LGA and JFK)?  float
q_4_a = ((flights_data['dest'] == 'SEA') & ((flights_data['origin'] == 'LGA') | (flights_data['origin'] == 'JFK'))).sum() #sum the occurances where the destination is SEA and the origin is a NY airport 
q_4 = float(q_4_a/(flights_data['dest'] == 'SEA').sum()) #create a fraction of those specific flights over all SEA flights
print(q_4)
```

```python
#Question 5 Which date has the largest average depature delay? Pd slice with date and float
#please make date a column. Preferred format is 2013/1/1 (y/m/d)
flights_data['date'] = (flights_data['year'].astype(str) + '/' + 
                        flights_data['month'].astype(str) + '/' +
                        flights_data['day'].astype(str)) #build the new column
delay_by_day = flights_data.groupby('date', as_index=False)['dep_delay'].mean() #index the dates by departure delay
q_5 = delay_by_day[delay_by_day['dep_delay'] == delay_by_day['dep_delay'].max()] #compare departure delays and keep the max
print(q_5) # display the dataframe slice with the date and max departure delay
```

```python
#Question 6 Which date has the largest average arrival delay? pd slice with date and float
delay_by_day = flights_data.groupby('date', as_index=False)['arr_delay'].mean() #index the dates by arrival delay
q_6 = delay_by_day[delay_by_day['arr_delay'] == delay_by_day['arr_delay'].max()] #compare arrival delays and keep the max
print (q_6)
```

```python
#Question 7 Which flight departing LGA or JFK in 2013 flew the fastest? pd slice with tailnumber and speed
#speed = distance/airtime
fastest_flight = flights_data[flights_data['origin'].isin(['LGA','JFK']) & (flights_data['year'] == 2013)].copy() #only include rows with right origin and year
fastest_flight['speed'] = fastest_flight['distance']/fastest_flight['air_time'] # calculate speed
q_7 = fastest_flight.loc[fastest_flight['speed'] == fastest_flight['speed'].max(), ['tailnum','speed']] #compare speeds among all the rows and select the row with the max speed, report only tailnum and speed columns
print(q_7)
```

```python
#Question 8 Replace all nans in the weather pd dataframe with 0s. Pd with no nans
q_8 = weather_data_pd.fillna(0)
```

```python
#%% Numpy Data Filtering/Sorting Question Answering
#Use weather_data_np
#Question 9 How many observations were made in Feburary? Int
q_9 = int((weather_data_np[:, 3] == 2).sum()) #count the true instances among all rows where column in index position 3 equals 2
print(q_9)
```

```python
#Question 10 What was the mean for humidity in February? Float
q_10 = float((weather_data_np[weather_data_np[:, 3] == 2, 8]).mean()) #Select rows where the month is february, and average the humidity column (index 8)
print(q_10)
```

```python
#Question 11 What was the std for humidity in February? Float
q_11 = float((weather_data_np[weather_data_np[:, 3] == 2, 8]).std()) #Select rows where the month is february, and calculate standard dev of the humidity column (index 8)
print(q_11)
```

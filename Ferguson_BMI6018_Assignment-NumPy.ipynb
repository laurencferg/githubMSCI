```python
from sklearn.datasets import load_iris

iris = load_iris()
```

```python
petalwidth = iris.data[:,3]
threshold = petalwidth > 1.0

if threshold.any():
    first_big_petal = np.argmax(threshold)
    print (f"The first instance of a petal width greater than 1.0 is at row {first_big_petal}, column 4")
else:
    print("There are no petals greater than 1.0")
```

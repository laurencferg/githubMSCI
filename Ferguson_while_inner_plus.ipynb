```python
def inner_list(n):
    candidates = [n] #entire input list comprises options for innermost list, initially
    innermost = n #set up eventual innermost winner
    while candidates: #run while there are options for innermost
        current = candidates.pop(0) #evaluate the first item in the candidates list, index 0
        is_nested = any(isinstance(item, list) for item in current) #set up evaluation of whether the item in current has a nested list
        if is_nested: #if current does contain a nested list...
            for item in current:
                if isinstance(item, list): #find the item in the list that IS a list 
                    candidates.append(item) #and add that item to the list of candidates to evaluate for deepest
        else:
            innermost = current #and if current does not contain a sublist, it is the innermost
    result_list = [x + 1 for x in innermost] #add 1 to each item in the innermost list
    print(result_list)
    return result_list

input_list = [1,2,3,4,[5,6,7,[8,9]]]
inner_list(input_list)
```

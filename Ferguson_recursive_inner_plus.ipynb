```python
def inner_list(n):
    def find_inner_list(n, depth=0): #an inner function prepared to track depth of the candidate innermost lists
        is_nested = any(isinstance(item, list) for item in n) #set up evaluation of whether the input contains a sublist
        
        if not is_nested: #base case, no sublists detected anymore
            return n, depth #report the innermost list and the depth at which it was found
            
        innermost, deepest = n, depth #label the current list as the innermost and its depth
        
        for item in n: #for each item in the current list
            if isinstance(item, list): #if that item is a list
                candidate_list, candidate_depth = find_inner_list(item, depth + 1) #repeat the function and label that output as a candidate innermost list
                if candidate_depth > deepest: #compare current lists depth to the reigning champion depth
                    innermost, deepest = candidate_list, candidate_depth #if the new depth is greater, allow that list and depth to take position as innermost
        return innermost, deepest #store these values for use outside the function

    innermost, _ = find_inner_list(n) #unpack the result outside of the recursive function
    result_list = [x + 1 for x in innermost] #add 1 to each item in the innermost list
    print(result_list)
    return result_list

input_list = [1,2,3,4,[5,6,7,[8,9]]]
inner_list(input_list)
```
